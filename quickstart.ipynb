```python
import sys
import os

import json
import random
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import torch
from torch.utils.data import DataLoader
from tqdm import tqdm
import cv2
from datasets.fmd.data_loader import load_noisy_single_loader
from QUTCC_eval import create_broken_axis_pdf_plot
from evaluation import get_largest_residual_indices
from evaluation.calibration_utils import (AnalysisManager, CalibrationManager,
                                          get_test_dataloader)
from models.checkpointing import load_checkpoint_for_inference, create_model, load_model_state
from QUTCC_eval import plot_visualization

np.random.seed(0)
random.seed(0)
torch.manual_seed(0)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
print("tmp folder: ", os.environ["TMPDIR"])

%matplotlib inline
%load_ext autoreload
%autoreload 2
```

```python
#Load in weights + calibration values
im2im_deep_path = "./QUTCC_eval/im2im_deep.pth" #Epoch 15
qutcc_path = "./QUTCC_eval/qutcc.pth" #Epoch 20

#Pre-calibrated values:
im2im_deep_lambda = 1.0910910367965698
upper_q = 0.9667335087433457
lower_q = 1e-20
quantile_levels = [0.05, 0.1, 0.15, 0.2, 0.25, 0.5, 0.75, 0.8, 0.85, 0.9, 0.95]
quantile_levels_conformal = [1e-20,0.06856392578125, 0.12036991210937498,0.164851689453125,0.20728262695312502,
                             0.5, 0.6923769140625 , 0.7546805664062501 , 0.8198651660156251, 0.8874912500000001,
                             0.9667335087433457]

#Load in clean/noisy image
noisy = plt.imread('./QUTCC_eval/noisy.png', -1).astype(np.float32)
ground_truth = plt.imread('./QUTCC_eval/ground_truth.png', -1).astype(np.float32)
#Load in tuned model checkpoint
im2im_deep = create_model(net = "unet_im2im", device = device)
im2im_deep = load_model_state(im2im_deep, im2im_deep_path, device)

qutcc = create_model(net = "unet_quantile", device = device)
qutcc = load_model_state(qutcc, qutcc_path, device)
```

```python
#Visualization
fig, ax = plot_visualization(noisy, ground_truth, im2im_model=im2im_deep, 
                    quantile_model=qutcc, im2im_lam=im2im_deep_lambda, lower_q=lower_q, upper_q=upper_q, 
                    device=device, residual_vmax=0.2, uncertainty_vmax=0.5, exp_type='Gaussian', save=False)
plt.show()
```
